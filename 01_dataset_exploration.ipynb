```python
import GEOparse
import pandas as pd
import numpy as np

gse = GEOparse.get_GEO(
    geo="GSE2034",
    destdir="../data/raw/",
    silent=True,
    how="quick"
)

print(gse)
```

```python
sample = list(gse.gsms.values())[0]

sample.table.head()
```

```python
expression = {}

for sample_id, sample in gse.gsms.items():
    expression[sample_id] = (
        sample.table
        .set_index("ID_REF")["VALUE"]
    )

expression_df = pd.DataFrame(expression)

expression_df.head()
```
